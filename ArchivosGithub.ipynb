# Función genérica

```python
def subir_carpeta_a_github(zip_path, patron, carpeta_destino, mensaje_commit, repo=repo):
    """
    zip_path: ruta al .zip subido a Colab (ej. '/content/gm_estandar.zip')
    patron: patrón glob que deben calzar los archivos válidos (ej. 'gm-sinader-*.csv') —
            cualquier otro archivo que venga en el zip (Thumbs.db, carpetas, etc.) se ignora
    carpeta_destino: subcarpeta dentro del repo (ej. 'data/Generacion_Municipal')
    mensaje_commit: mensaje del commit
    """
    nombre = os.path.splitext(os.path.basename(zip_path))[0]
    carpeta_temp = f"/content/_tmp_{nombre}"
    os.makedirs(carpeta_temp, exist_ok=True)

    !unzip -q -j "{zip_path}" -d "{carpeta_temp}"

    destino = f"{repo}/{carpeta_destino}"
    os.makedirs(destino, exist_ok=True)

    archivos = sorted(glob.glob(f"{carpeta_temp}/{patron}"))
    print(f"'{zip_path}': {len(archivos)} archivos calzan con '{patron}' (de {len(os.listdir(carpeta_temp))} en el zip)")
    for f in archivos:
        shutil.copy(f, destino)

    en_destino = sorted(glob.glob(f"{destino}/{patron}"))
    print(f"En {carpeta_destino}: {len(en_destino)} archivos")

    !git -C "{repo}" add "{carpeta_destino}"
    !git -C "{repo}" commit -m "{mensaje_commit}"
    !git -C "{repo}" push

    shutil.rmtree(carpeta_temp)
    return en_destino
```
